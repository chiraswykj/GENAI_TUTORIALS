```python
import os
import getpass

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

print("Libraries imported successfully!")
```

```python
os.environ["GROQ_API_KEY"] = getpass.getpass(
    "Enter your Groq API key: "
)

print("API key configured!")
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

print("Groq model initialized!")
```

```python
response = llm.invoke(
    "Explain Artificial Intelligence in simple words."
)

print(response.content)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", """
    You are an AI Research Assistant.

    Generate a well-structured research report
    based on the topic provided by the user.

    Include:
    1. Introduction
    2. Key Concepts
    3. Applications
    4. Advantages
    5. Disadvantages
    6. Future Scope
    7. Conclusion

    Use simple and informative language.
    """),

    ("human", "Research Topic: {topic}")
])

print("Research prompt created!")
```

```python
research_chain = prompt | llm

print("Research chain created successfully!")
```

```python
topic = "Artificial Intelligence in Healthcare"

result = research_chain.invoke({
    "topic": topic
})

print(result.content)
```

```python
def research_assistant():

    topic = input("Enter your research topic: ")

    print("\nGenerating research report...\n")

    result = research_chain.invoke({
        "topic": topic
    })

    print("=" * 50)
    print("AI RESEARCH REPORT")
    print("=" * 50)

    print(result.content)


research_assistant()
```
