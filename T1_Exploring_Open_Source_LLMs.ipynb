```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
output = generator(
    "Artificial intelligence is",
    max_new_tokens=30
)
print(output)
```

```python
output = generator(
    "Artificial intelligence is",
    max_new_tokens=10
)
print(output)
```

```python
output = generator(
    "Artificial intelligence is",
    max_new_tokens=50
)
print(output)
```

```python
from transformers import AutoTokenizer
```

```python
tokenizer=AutoTokenizer.from_pretrained("gpt2")
```

```python
text="Generative AI is transforming education"
```

```python
tokens=tokenizer.tokenize(text)
print(tokens)
```

```python
token_ids=tokenizer.encode(text)
print(token_ids)
```
